```python
import os
from langchain_core.prompts import ChatPromptTemplate
from langchain_openai import ChatOpenAI
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnableLambda
```

```python
def deixar_maiusculo(texto: str) -> str:
    return texto.upper()
```

```python
prompt = ChatPromptTemplate.from_template("explique em uma frase o {assunto}")

model = ChatOpenAI(model="gpt-5-nano-2025-08-07", openai_api_key=os.getenv("OPENAI_API_KEY"))

parser = StrOutputParser()

passo_extra = RunnableLambda(deixar_maiusculo)
```

```python
pipeline = prompt | model | parser | passo_extra
```

```python
resultado = pipeline.invoke({"assunto": "inteligência artificial"})
print(resultado)
```
